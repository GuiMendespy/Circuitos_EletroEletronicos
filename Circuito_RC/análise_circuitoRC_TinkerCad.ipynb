# Análise de Circuito RC

Programa em Python para calcular a tensão no capacitor (Vc),
tensão no resistor (Vr) e corrente (i) durante os processos
de carga e descarga de um circuito RC.

## Parâmetros

- R = 10 kΩ
- C = 10 µF
- E = 10 V
- f = 1 Hz

## Constante de tempo

τ = RC = 0,1 s

## Período

T = 1/f = 1 s

Como a fonte possui duty cycle de 50%, o capacitor possui
0,5 s para carregar e 0,5 s para descarregar.

## Equações utilizadas

### Carga

Vc(t) = E(1 - e^(-t/RC))

Vr(t) = E - Vc(t)

i(t) = Vr(t)/R

### Descarga

Vc(t) = Vc0 e^(-t/RC)

Vr(t) = -Vc(t)

i(t) = Vr(t)/R

```python
import math


# ==========================
# PARÂMETROS DO CIRCUITO
# ==========================

R = 10_000       # Resistência em ohms (10 kΩ)
C = 10e-6        # Capacitância em farads (10 µF)
E = 10           # Tensão da fonte em volts (10 V)
f = 1            # Frequência em Hz

# ==========================
# CÁLCULOS INICIAIS
# ==========================

tau = R * C       # Constante de tempo
T = 1 / f         # Período
tempo_meio = T / 2

print("=" * 65)
print("         CIRCUITO RC - CARGA E DESCARGA")
print("=" * 65)

print(f"R = {R / 1000:.1f} kΩ")
print(f"C = {C * 1e6:.1f} µF")
print(f"E = {E:.1f} V")
print(f"f = {f:.1f} Hz")
print(f"τ = {tau:.3f} s")
print(f"T = {T:.3f} s")
print(f"Tempo de carga/descarga = {tempo_meio:.3f} s")

# ==========================
# TABELA DE CARGA
# ==========================

print("\n")
print("=" * 65)
print("                      CARGA")
print("=" * 65)

print(f"{'Tempo (s)':>12} {'t/τ':>10} {'Vc (V)':>12} "
      f"{'Vr (V)':>12} {'i (mA)':>12}")

print("-" * 65)

for n in range(6):

    t = n * tau

    # Tensão no capacitor
    Vc = E * (1 - math.exp(-t / tau))

    # Tensão no resistor
    Vr = E - Vc

    # Corrente
    i = Vr / R

    print(f"{t:>12.3f} {n:>10} {Vc:>12.2f} "
          f"{Vr:>12.2f} {i * 1000:>12.3f}")

# ==========================
# TABELA DE DESCARGA
# ==========================

# Tensão no capacitor no início da descarga
Vc0 = E * (1 - math.exp(-tempo_meio / tau))

print("\n")
print("=" * 65)
print("                    DESCARGA")
print("=" * 65)

print(f"{'Tempo (s)':>12} {'t/τ':>10} {'Vc (V)':>12} "
      f"{'Vr (V)':>12} {'i (mA)':>12}")

print("-" * 65)

for n in range(6):

    t = n * tau

    # Tensão no capacitor durante a descarga
    Vc = Vc0 * math.exp(-t / tau)

    # Tensão no resistor
    Vr = -Vc

    # Corrente
    i = Vr / R

    print(f"{t:>12.3f} {n:>10} {Vc:>12.2f} "
          f"{Vr:>12.2f} {i * 1000:>12.3f}")

print("\n" + "=" * 65)
print("Fim da análise.")
print("=" * 65)
```
