```python
import math

# Parâmetros
R = 100.0        # ohms
L = 0.1          # henry (100 mH)
E = 10.0         # volts
f = 1.0          # Hz

tau = L / R      # constante de tempo
T = 1 / f
Ton = T / 2      # 0.5 s
Toff = T / 2     # 0.5 s

print("\nCIRCUITO RL - CARGA E DESCARGA")
print(f"R = {R} Ω, L = {L} H, E = {E} V, f = {f} Hz")
print(f"tau = {tau} s = {tau*1000:.3f} ms\n")

# ----- TABELA DE CARGA -----
print("TABELA - CARGA (0 a 10 V)")
print(f"{'t (s)':>8} {'t/tau':>8} {'i (A)':>10} {'VR (V)':>10} {'VL (V)':>10}")
print('-'*50)
tempos_carga = [0, 1, 2, 3, 4, 5, 10, 50]
for n in tempos_carga:
    t = n * tau
    i = (E/R) * (1 - math.exp(-t/tau))
    VR = R * i
    VL = E * math.exp(-t/tau)
    print(f"{t:8.4f} {n:8.0f} {i:10.5f} {VR:10.5f} {VL:10.5f}")

# ----- TABELA DE DESCARGA -----
print("\nTABELA - DESCARGA (10 V -> 0 V)")
print(f"{'t abs (s)':>10} {'t desc (s)':>11} {'t/tau':>8} {'i (A)':>10} {'VR (V)':>10} {'VL (V)':>10}")
print('-'*65)
tempos_descarga = [0, 1, 2, 3, 4, 5, 10, 50]
for n in tempos_descarga:
    t_desc = n * tau
    t_abs = Ton + t_desc
    i = (E/R) * math.exp(-t_desc/tau)
    VR = R * i
    VL = -R * i
    print(f"{t_abs:10.3f} {t_desc:11.3f} {n:8.0f} {i:10.6f} {VR:10.6f} {VL:10.6f}")
```
